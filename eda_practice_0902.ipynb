결측 처리 1

```python
missing_cols = ['city', 'winner', 'player_of_match', 'umpire1', 'umpire2']
```

```python
for col in missing_cols:
    df[col].value_counts().plot(kind = 'bar', figsize = (10, 5))
    plt.title(f'{col} value counts plot')
    plt.tight_layout()
    plt.show()
```

```python
'''
결측치 수가 적기 때문에 그냥 제거 하는 것으로 함
'''

df.dropna(subset = missing_cols, inplace = True)
```

```python
numeric_df = df.select_dtypes(include = ['int64', 'float64'])
object_df = df.select_dtypes(include = ['object'])

for col in numeric_df.columns:
    plt.figure(figsize = (6, 3))
    sns.histplot(numeric_df[col], kde = True)
    plt.title(f'{col} histogram plot')
    plt.tight_layout()
    plt.show()

for col in object_df.columns:
    df[col].value_counts().plot(kind = 'bar', figsize = (10, 5))
    plt.title(f'{col} value counts plot')
    plt.ylabel('count')
    plt.show()
    plt.tight_layout()

```
